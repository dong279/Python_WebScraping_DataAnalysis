### 4-1. 인구현황 데이터 분석


```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 한글 폰트 설정 (없으면 축 라벨이 □□□ 로 깨진다)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

pop_data = pd.read_csv('../data/인구현황.csv')
print(pop_data.shape)

# 첫 행은 '전국' 합계이므로 지역별 분석에서 제외한다.
pop_data = pop_data[pop_data['행정기관'] != '전국']
print(pop_data.shape)

pop_data.head()
```

### 1. 지역별 총인구수 Top5


```python
top5_pop = pop_data[['행정기관', '총인구수']] \
    .sort_values('총인구수', ascending=False) \
    .head(5)

top5_pop
```

### 1-1. 지역별 총인구수 plot

```python
plot_data = pop_data.sort_values('총인구수', ascending=False)

plt.figure(figsize=(12, 5))
sns.barplot(data=plot_data, x='행정기관', y='총인구수', hue='행정기관')
plt.title('지역별 총인구수')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### 2. 세대당 인구수 높은 지역 Top5


```python
top5_household = pop_data[['행정기관', '세대당 인구']] \
    .sort_values('세대당 인구', ascending=False) \
    .reset_index(drop=True) \
    .head(5)

top5_household
```

### 2-1. 지역별 세대당 인구 Plot


```python
plot_data = pop_data.sort_values('세대당 인구', ascending=False)

plt.figure(figsize=(12, 5))
sns.barplot(data=plot_data, x='행정기관', y='세대당 인구', hue='행정기관')
plt.title('지역별 세대당 인구')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### 3. 남녀 비율 분석


```python
ratio_data = pop_data[['행정기관', '남여 비율']] \
    .sort_values('남여 비율', ascending=False) \
    .reset_index(drop=True)

ratio_data
```

### 3-1. 남녀 비율 분석 Plot


```python
plt.figure(figsize=(12, 5))
sns.barplot(data=ratio_data, x='행정기관', y='남여 비율', hue='행정기관')

# 기준이 되는 빨간선 : 남녀 1:1 비율
plt.axhline(y=1.0, color='red', linestyle='--', label='남녀 1:1 비율')

plt.title('지역별 남녀 비율')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```

### 4. 남자초과 / 여자초과 분석


```python
def classify_gender(ratio):
    """남여 비율로 남초/여초/동일을 판별한다."""
    if ratio > 1:
        return '남초'
    elif ratio < 1:
        return '여초'
    else:
        return '동일'


pop_data['남초여초'] = pop_data['남여 비율'].apply(classify_gender)

gender_data = pop_data[['행정기관', '남여 비율', '남초여초']]
gender_data
```

### 5. 세대당 인구 평균보다 높은 지역

```python
mean_household = round(pop_data['세대당 인구'].mean(), 2)

print('5. 세대당 인구 평균보다 높은 지역')
print(f'전국 평균 세대당 인구수: {mean_household}')

above_mean = pop_data[pop_data['세대당 인구'] > mean_household][
    ['행정기관', '세대당 인구']] \
    .sort_values('세대당 인구', ascending=False) \
    .reset_index(drop=True)

above_mean
```

### 5-1. 남초 vs 여초 개수 Plot  ( seaborn의 countplot )


```python
plt.figure(figsize=(8, 4))
sns.countplot(data=pop_data, x='남초여초', hue='남초여초')
plt.title('남초 vs 여초 지역 개수')
plt.tight_layout()
plt.show()
```

###  5-1. 지역별 세대수 Plot ( seaborn의 barplot )

```python
plot_data = pop_data.sort_values('세대수', ascending=False)

plt.figure(figsize=(14, 5))
sns.barplot(data=plot_data, x='행정기관', y='세대수', hue='행정기관')
plt.title('지역별 세대수')
plt.tight_layout()
plt.show()
```
