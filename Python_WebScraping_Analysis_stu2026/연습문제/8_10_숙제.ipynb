### 1-1. Daum 뉴스기사 제목 스크래핑하기 


```python
import requests
from bs4 import BeautifulSoup


# 다음 경제 뉴스 URL
url = 'https://news.daum.net/economy'

# 요청 헤더 설정 : 브라우저 정보
req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                  'AppleWebKit/537.36 (KHTML, like Gecko) '
                  'Chrome/144.0.0.0 Safari/537.36'
}

res = requests.get(url, headers=req_header)
res.encoding = 'utf-8'

print(url)
print(type(res))
print(res.status_code)

if res.ok:
    html = res.text
    soup = BeautifulSoup(html, 'html.parser')

    li_tags = soup.select('ul.list_newsheadline2 li')
    print(type(li_tags), len(li_tags))

    for li_tag in li_tags:
        a_tag = li_tag.find('a')
        link = a_tag['href']

        strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
        title = strong_tag.text.strip()

        print(link)
        print(title)
else:
    print(f'Error Code = {res.status_code}')
```

### 1-2:  여러개의 section 중 하나를 선택해서 url에서 뉴스기사의 링크와 제목을 출력하는 코드를 함수로 작성하기


```python
import requests
from bs4 import BeautifulSoup


section_dict = {
    '기후/환경': 'climate',
    '사회': 'society',
    '경제': 'economy',
    '정치': 'politics',
    '국제': 'world',
    '문화': 'culture',
    '생활': 'life',
    'IT/과학': 'tech',
    '인물': 'people',
}


def print_news(section_name):
    """Daum 뉴스 섹션의 기사 링크와 제목을 출력한다."""
    section_code = section_dict.get(section_name, 'economy')
    url = f'https://news.daum.net/{section_code}'
    print(f'======> {url} {section_name} 뉴스 <======')

    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                      'AppleWebKit/537.36 (KHTML, like Gecko) '
                      'Chrome/144.0.0.0 Safari/537.36'
    }

    res = requests.get(url, headers=req_header)
    res.encoding = 'utf-8'

    if res.ok:
        html = res.text
        soup = BeautifulSoup(html, 'html.parser')

        for li_tag in soup.select('ul.list_newsheadline2 li'):
            a_tag = li_tag.find('a')
            link = a_tag['href']

            strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
            if strong_tag is None:
                continue

            print(link)
            print(strong_tag.text.strip())
    else:
        print(f'Error Code = {res.status_code}')


print_news('경제')
print_news('사회')
```

### 2-1. Nate 뉴스 스크래핑

```python
from urllib.parse import urljoin

import requests
from bs4 import BeautifulSoup
from IPython.display import Image, display


nate_dict = {
    '최신뉴스': 'n0100',
    '정치': 'n0200',
    '경제': 'n0300',
    '사회': 'n0400',
    '세계': 'n0500',
    'IT/과학': 'n0600',
}


def print_nate_news(section_name):
    """Nate 뉴스 섹션의 이미지, 제목, 링크를 출력한다."""
    mid = nate_dict.get(section_name, 'n0100')
    url = f'https://news.nate.com/recent?mid={mid}'
    print(f'======> {url} {section_name} 뉴스 <======')

    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                      'AppleWebKit/537.36 (KHTML, like Gecko) '
                      'Chrome/144.0.0.0 Safari/537.36'
    }

    res = requests.get(url, headers=req_header)
    res.encoding = 'euc-kr'

    if res.ok:
        html = res.text
        soup = BeautifulSoup(html, 'html.parser')

        for item in soup.select('div.mduSubject div.mlt01'):
            a_tag = item.find('a')
            link = urljoin(url, a_tag['href'])

            title_tag = item.select_one('strong.tit')
            if title_tag is None:
                continue

            print(link)
            print(title_tag.text.strip())

            # 이미지가 없는 뉴스도 있으므로 존재 여부를 확인한다.
            img_tag = item.find('img')
            if img_tag is not None:
                img_url = urljoin(url, img_tag['src'])
                display(Image(url=img_url))

            print('-' * 60)
    else:
        print(f'Error Code = {res.status_code}')


for name in nate_dict:
    print_nate_news(name)
```

### 2-2. 네이버 웹툰 이미지 다운로드

```python
import os

import requests
from bs4 import BeautifulSoup


def download_one_episode(title, no, url):
    """웹툰 한 회차의 이미지를 img/제목/회차 디렉토리에 저장한다."""
    # referer 요청 헤더를 반드시 설정해야 한다.
    req_header = {
        'referer': url,
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                      'AppleWebKit/537.36 (KHTML, like Gecko) '
                      'Chrome/144.0.0.0 Safari/537.36'
    }

    # 이미지 저장 폴더가 없으면 생성 ( sub 디렉토리도 생성 )
    imgdir_name = os.path.join('img', title, str(no))
    os.makedirs(imgdir_name, exist_ok=True)
    print(f'저장 경로 = {imgdir_name}')

    res = requests.get(url, headers=req_header)
    if not res.ok:
        print(f'Error Code = {res.status_code}')
        return

    # 이미지 URL 추출 - List Comprehension
    soup = BeautifulSoup(res.text, 'html.parser')
    img_tags = soup.select("img[src*='IMAG01']")
    img_url_list = [img_tag['src'] for img_tag in img_tags]
    print(f'이미지 개수 = {len(img_url_list)}')

    # 이미지 다운로드
    for img_url in img_url_list:
        res = requests.get(img_url, headers=req_header)
        if res.ok:
            img_data = res.content
            file_path = os.path.join(imgdir_name,
                                     os.path.basename(img_url))
            with open(file_path, 'wb') as file:
                print(f'Writing to {file_path} ({len(img_data):,} bytes)')
                file.write(img_data)
        else:
            print(f'Error Code = {res.status_code} for {img_url}')


download_one_episode(
    '일렉시드',
    341,
    'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed',
)
```
